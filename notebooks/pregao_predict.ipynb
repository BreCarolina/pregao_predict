```python
#@title Objetivo do Projeto

%%writefile /content/MVP_Objetivo.md

### 1. Propósito do MVP

Este projeto tem como objetivo principal a criação de um pipeline para extração, transformação, carga, análise e previsão da movimentação intradiária dos preços de um ativo financeiro em intervalos de 5 minutos.
O modelo preditivo central será baseado em redes neurais recorrentes (LSTM), mas outras abordagens serão exploradas.
O MVP visa garantir previsões para embasar decisões estratégicas de day trade.

### 2. Problema a Ser Resolvido

A alta volatilidade dos mercados financeiros exige ferramentas robustas para antecipação de movimentos de preço.
A dificuldade está em capturar padrões de curto prazo e projetá-los com precisão.
Traders e investidores necessitam de um modelo que consiga interpretar os padrões históricos e transformá-los em previsões úteis.

### 3. Pipeline do Projeto

O projeto será estruturado em sete etapas:

1. **Extração e armazenamento dos dados brutos:**
Coleta de dados históricos do ativo em questao, utilizando a API do Yahoo Finance (yfinance).
Os dados serão armazenados em um repositório GitHub sincronizado com Google Colab, garantindo acesso remoto e backup na nuvem.

2. **Limpeza e organização dos dados:** Padronização de colunas, tratamento de dados ausentes, eliminação de duplicatas e organização cronológica.
Resultado salvo como `dados_limpos.csv`.

3. **Transformação e engenharia de features:** Adição de indicadores técnicos e features temporais.
Resultado salvo como `dados_transformados.csv`.

4. **Modelagem e estruturação do banco de dados:**

   a) Organização em arquivos CSV:

   - `dados_brutos.csv`: dados originais extraídos da API.
   - `dados_limpos.csv`: após limpeza e padronização.
   - `dados_transformados.csv`: após adição de features técnicas.
   - `dados_preparados.csv`: versão padronizada e normalizada dos dados dividida em treino teste e validação.

   b) Banco de dados dimensional:
      Modelagem de dados no esquema estrela para facilitar a análise de fatos.

   - **Fato**: `fato_precos`, contendo os valores de fechamento e chaves para dimensões.
   - **Dimensões**:
     - `dim_tempo`: atributos temporais.
     - `dim_indicadores`: indicadores técnicos.
     - `dim_agregados`: dados agregados para enriquecer a análise.
     - `dim_operacional': dados operacionais.

   c) Catalogo de dados

  - Um **Catálogo de Dados** será elaborado com descrição, domínio, categorias e linhagem de cada variável.

5. **Carga e Pipeline ETL:** Pipeline estruturado com as seguintes etapas:

   - **Extração:** via API do Yahoo Finance.
   - **Limpeza:** tratamento e estruturação básica.
   - **Transformação:** geração de variáveis técnicas e derivadas.
   - **Carga:** integração dos dados transformados no banco de dados dimensional.

6. **Treinamento e ajuste do modelo:** Implementação e ajuste do modelo preditivo com base LSTM, com avaliação por métricas como MSE e R².

7. **Interpretação dos resultados e resposta às perguntas:** Validação das previsões.

### 4. Perguntas a Serem Respondidas

01 - Qual horário do dia é comum maior alta do ativo?
02 - Existe algum padrão proximo ao horário de fechamento do pregão?
03 - Qual progressão média do ativo por mes?
04 - Com qual volume de negociações normalmente temos uma inversão?
05 - Qual correlação do RSI com a inversão?
06 - Qual correlação do RSI com o preço de fechamento?
07 - É possivel gerar a previsão completa do pregão de 5 em 5 minutos com precisão?
08 - É possivel prever as targets globais do ativo no pregao? (Abertura, fechamento, minima e máxima)
09 - Qual tendencia de preço do ativo em um ano?
10 - Existe algum padrao que constitui a queda do preço do ativo?
11 - Os dados do dia anterior fornecem informações suficientes para a previsão do dia seguinte?
12 - A modelagem com LSTM captura corretamente as tendências de curto prazo?
13 - Quais indicadores técnicos e features são mais relevantes para melhorar a acurácia do modelo?
14 - Como considerar corretamente as quebras de fim de semana?
15 - A normalização e padronização das variáveis melhora a precisão do modelo?

### 5. Critérios de Sucesso

Para que o MVP seja considerado bem-sucedido o esperado é que:

1. O pipeline de extração, transformação e previsão funcione de forma eficiente.
2. O modelo consiga prever a movimentação dos preços com um erro médio aceitável.
3. A previsão de targets globais (abertura, máxima, mínima, fechamento) seja consistente com os valores reais em pelo menos 90%.
4. O modelo consiga lidar corretamente com fins de semana e feriados.
5. As previsões sejam suficientes para auxiliar na tomada de decisão de trading.

```

```python
#@title Readme.md
%%writefile /content/README.md


# Pipeline de Processamento de Dados de Ativos da B3

Este projeto realiza a extração, limpeza, transformação e carga incremental de dados financeiros de ativos da B3 (ex: PETR4, BBDC4, LREN3), estruturando-os em um banco de dados dimensional com modelo estrela.

O objetivo final é utilizar esses dados como base para desenvolvimento de modelos preditivos, especialmente modelos LSTM (Long Short-Term Memory), voltados à previsão de preços e comportamentos do mercado intradiário. A etapa de modelagem ainda está em desenvolvimento.

Este projeto é disponibilizado sob uma **licença personalizada de uso restrito**. Consulte a seção "Licença" ao final deste documento para mais detalhes.

---

## Ambiente de Execução

* O projeto foi concebido para execução em ambiente **Google Colab**, aproveitando a estrutura de diretórios padrão `/content/`.
* Certifique-se de montar corretamente a árvore de pastas (`Piloto_day_trade/`) dentro do ambiente Colab.
* Variáveis de ambiente sensíveis devem ser configuradas via `.env` (utilizando `python-dotenv`).

---

## Controle de Versão

O código-fonte está versionado utilizando **Git**. Recomenda-se clonar este repositório via GitHub para manter a rastreabilidade de alterações e contribuir com melhorias:

!git clone https://github.com/CarolBw/Piloto_day_trade.git


## Estrutura do Projeto

Piloto_day_trade/
├── data/
│   ├── raw/                  # Dados brutos extraídos (yfinance)
│   ├── cleaned/              # Dados limpos após filtragem de horários e duplicatas
│   ├── prepared/             # Reservado para features intermediárias
│   └── transformed/          # Dados finais prontos para carga
├── docs/                     # Documentações auxiliares
├── modelagem/
│   ├── catalog/              # Dicionário de dados e catálogos
│   ├── database/             # Estrutura e arquivos do banco
│   └── esquema/              # Modelo estrela e relações
├── models/                   # Modelos treinados ou salvos
├── notebooks/                # Análises exploratórias e estudos
├── reports/                  # Relatórios e saídas de modelos
│   └── figures/              # Gráficos gerados
├── src/
│   ├── modelagem_ml/         # Scripts de modelagem e previsão
│   ├── operacional/          # Utilitários operacionais
│   ├── pipeline/
│   │   ├── transformacao_dados.py  # Enriquecimento com indicadores e features
│   │   ├── carga_dados.py         # Carga incremental no SQLite
│   │   └── pipeline_<ticker>.py   # Pipeline ETL completo para um ticker
│   └── extrair_dados.py      # Extração e limpeza
├── tcc/                      # Documentos de TCC
├── README.md
├── requirements.txt
├── .env
├── .gitignore
└── LICENSE

## Etapas do Pipeline

### 1. Extração e Limpeza (`extrair_dados.py`)

* Baixa dados intradiários de ativos via `yfinance`.
* Converte datas e horários para timezone BRT.
* Renomeia colunas para nomes padronizados (`abertura`, `minimo`, `maximo`, `fechamento`, `volume`).
* Remove nulos, duplicatas e horários fora do pregão (09:55 - 18:05).
* Salva em `data/cleaned/`.

### 2. Transformação (`transformacao_dados.py`)

* Carrega os dados limpos e incrementa apenas registros ainda não processados.
* Ordena por data e hora.
* Calcula indicadores técnicos:

  * Médias móveis (SMA, EMA)
  * MACD e linha de sinal
  * RSI
  * OBV (On-Balance Volume)
  * Log-retorno e volatilidade
* Adiciona atributos temporais:

  * `dia_da_semana`, `hora_num`, `minuto`
* Adiciona estatísticas diárias (atuais e do dia anterior):

  * Fechamento, volume, máximo, mínimo
* Preenche nulos técnicos com a média da respectiva coluna.
* Salva em `data/transformed/`.

### 3. Carga Dimensional (`carga_dados.py`)

* Lê os dados transformados e conecta-se ao SQLite (`modelagem/database/`).
* Cria automaticamente as tabelas do modelo estrela, se não existirem:

  * **dim\_tempo**: data, hora, dia\_da\_semana
  * **dim\_indicadores**: indicadores técnicos e volume
  * **dim\_agregados**: horário numérico e estatísticas do dia anterior
  * **fato\_precos**: preços com referências às três dimensões
* Verifica se o par (`data`, `hora`) já foi carregado para evitar duplicações.
* Utiliza inserções condicionais (`INSERT OR IGNORE`).
* Garante integridade referencial entre fato e dimensões.
* Exibe a contagem de novos registros inseridos.

---

## Requisitos

* Python 3.10+
* Bibliotecas utilizadas no pipeline:

  * pandas, numpy, yfinance, matplotlib, seaborn
  * sqlite3, sqlalchemy
  * python-dotenv, gitpython

Instalação:

!pip install -r requirements.txt

---

## Execução

**Pipeline completo para um ativo**:

!python src/pipeline/pipeline_<ticker>.py


Substitua `<ticker>` por `BBDC4`, `LREN3`, etc. conforme desejado.

> **Importante:** Antes de executar o script principal do pipeline (ex: `pipeline_BBDC4.py`), verifique e ajuste os seguintes parâmetros no início do arquivo:
>
> * `ticker`: o código completo do ativo (ex: `BBDC4.SA`)
> * `intervalo`: intervalo de coleta dos dados (ex: `5m`, `15m`)
> * `dias`: número de dias de dados a serem buscados
> * Caminhos de entrada/saída dos dados e do banco SQLite, se necessário
>
> Esses parâmetros devem refletir suas necessidades específicas de coleta, transformação e armazenamento dos dados.

---

## Observações

* O pipeline é genérico e funciona para qualquer ticker da B3 aceito pelo Yahoo Finance (ex: `PETR4.SA`, `BBDC4.SA`, `LREN3.SA`).
* A transformação é incremental: evita retrabalho e reprocessamentos desnecessários.

---

## Em Desenvolvimento

* Criação do modelo preditivo LSTM para previsão de preços: estruturação, treinamento e avaliação
* Integração do pipeline com dashboard de visualização e monitoramento de dados e previsões
* Validações com testes automatizados
* Suporte a exportação em Parquet e bancos em nuvem

---

## 🔒 Licença

Copyright (c) 2025 Carolina Brescowitt

Todos os direitos reservados.

Este software é fornecido gratuitamente apenas para uso **pessoal, acadêmico e de pesquisa**.

O uso comercial deste software é estritamente proibido sem uma **licença comercial paga**, a ser negociada com o autor.

Empresas, startups, desenvolvedores ou qualquer entidade que deseje utilizar este código em produtos, serviços ou plataformas comerciais devem entrar em contato com o autor para **negociar os termos de licenciamento** (incluindo percentual, royalties ou valores fixos).

É proibida a redistribuição ou sublicenciamento sem autorização por escrito.

Para mais informações, entre em contato através do e-mail: [carolbrescowitt@yahoo.com.br](mailto:carolbrescowitt@yahoo.com.br)

---

Mantenha seus dados financeiros organizados e prontos para análise com este pipeline modular e extensível.
```

```python
%%writefile /content/configurar_git.py
#@title Script operacional para configurar Git e sincronizar com GitHub
import os
from dotenv import load_dotenv

load_dotenv(dotenv_path='/content/.env')

def git_config():
    """Configura o Git localmente e sincroniza com o repositório remoto no GitHub."""

    # Carregar variáveis do ambiente
    load_dotenv(dotenv_path='/content/.env')

    GITHUB_USERNAME = os.getenv('GITHUB_USERNAME')
    EMAIL = os.getenv('EMAIL')
    GITHUB_TOKEN = os.getenv('GITHUB_TOKEN')
    PROJECT_NAME = os.getenv('PROJECT_NAME')

    if not all([GITHUB_USERNAME, EMAIL, GITHUB_TOKEN, PROJECT_NAME]):
        raise ValueError("Variáveis de ambiente faltando. Verifique o arquivo .env.")

    REPO_URL = f"https://{GITHUB_USERNAME}:{GITHUB_TOKEN}@github.com/{GITHUB_USERNAME}/{PROJECT_NAME}.git"

    # Configurações globais do Git
    os.system(f'git config --global user.name "{GITHUB_USERNAME}"')
    os.system(f'git config --global user.email "{EMAIL}"')

    if os.path.isdir(PROJECT_NAME):
        print(f"[INFO] Diretório '{PROJECT_NAME}' já existe. Sincronizando...")
        os.chdir(PROJECT_NAME)

        os.system("git init")
        os.system("git remote remove origin || true")
        os.system(f"git remote add origin {REPO_URL}")
        os.system("git fetch origin")
        os.system("git checkout -B main")
        os.system("git pull origin main --allow-unrelated-histories --no-rebase")
    else:
        print(f"[INFO] Clonando o repositório '{PROJECT_NAME}'...")
        os.system(f"git clone {REPO_URL}")
        os.chdir(PROJECT_NAME)

    print(f"[SUCESSO] Git configurado e sincronizado com: {REPO_URL}")

if __name__ == "__main__":
    git_config()
```

# Pipeline ETL

```python
%%writefile /content/Piloto_day_trade/src/pipeline_etl/extracao_dados.py

# @title Extração e limpeza de dados

import os
import logging
import warnings
from datetime import datetime, timedelta

import pandas as pd
import yfinance as yf

# Configurações
warnings.simplefilter(action='ignore', category=FutureWarning)
logging.getLogger("yfinance").setLevel(logging.CRITICAL)

#--------------------------------------------------------------------------------------#

def limpeza_dados(df):
    print('\nIniciando limpeza de dados...')

    df.index = pd.to_datetime(df.index, utc=True).tz_convert("America/Sao_Paulo").tz_localize(None)

    df['hora'] = df.index.time
    df.index.name = 'datetime'
    df = df.reset_index()
    df['data'] = pd.to_datetime(df['datetime'].dt.date)

    print("\nAmostra após criação das colunas hora e data:")
    print(df.head(4))

    mapeamento_colunas = {
        'Close': 'fechamento',
        'High': 'maximo',
        'Low': 'minimo',
        'Open': 'abertura',
        'Volume': 'volume'
    }
    df.rename(columns=mapeamento_colunas, inplace=True)

    for col in ['abertura', 'minimo', 'maximo', 'fechamento']:
        df[col] = pd.to_numeric(df[col], errors='coerce').round(2)
    df['volume'] = pd.to_numeric(df['volume'], errors='coerce', downcast='integer')

    df = df[['data', 'hora', 'abertura', 'minimo', 'maximo', 'fechamento', 'volume', 'datetime']]

    print("\nAmostra após reorganizar as colunas:")
    print(df.head(2))

    df = df.drop_duplicates(keep='first')
    linhas_removidas = df[df.isna().sum(axis=1) > (df.shape[1] * 0.5)]
    df = df.dropna(thresh=df.shape[1] * 0.5)
    print(f"Removendo {len(linhas_removidas)} linhas com mais de 50% de NaN")
    if not linhas_removidas.empty:
        print("Datetimes removidos:")
        print(linhas_removidas['datetime'].tolist())

    print("\nAmostra após remover duplicatas e nulos:")
    print(df.head(2))

    df = df[df['data'].dt.weekday < 5]
    print("\nAmostra após filtrar apenas dias úteis:")
    print(df.head(2))

    df = df[(df['hora'] >= datetime.strptime('10:00', '%H:%M').time()) &
            (df['hora'] <= datetime.strptime('16:55', '%H:%M').time())]
    print("\nAmostra após filtro de horário (10:00-16:55):")
    print(df.head(4))

    if df.empty:
        print("O DataFrame ficou vazio após o filtro de horário. Verifique o intervalo de dados.")
    else:
        print("\nLimpeza de dados concluída com sucesso.\n")

    df = df.sort_values(["data", "hora"])
    print("\nAmostra de dados limpos e ordenados:")
    print(df.head(5))
    print(df.tail(5))

    return df

#--------------------------------------------------------------------------------------#

def extrair_dados(ticker, intervalo, path_projeto, dias_janela=58):
    print('\nExtração de dados...\n')

    path_dados_brutos = os.path.join(path_projeto, f'data/raw/{ticker.split(".")[0]}/dados_brutos_{ticker.split(".")[0]}.csv')
    path_dados_limpos = os.path.join(path_projeto, f'data/cleaned/{ticker.split(".")[0]}/dados_limpos_{ticker.split(".")[0]}.csv')

    df_total = pd.DataFrame()
    agora = datetime.now()
    hoje = agora.date()
    ontem = hoje - timedelta(days=1)

    data_fim = hoje if agora.hour >= 17 and hoje.weekday() < 5 else ontem
    data_inicio = datetime.combine(hoje - timedelta(days=dias_janela), datetime.min.time()).replace(hour=10, minute=0)

    primeira_extracao = True
    df_existente = pd.DataFrame()

    if os.path.exists(path_dados_limpos):
        df_existente = pd.read_csv(path_dados_limpos)
        df_existente["data"] = pd.to_datetime(df_existente["data"], errors="coerce")
        if "data" in df_existente.columns and pd.notnull(df_existente["data"].max()):
            ultimo_registro = df_existente["data"].max().date()
            print(f"Último registro encontrado: {ultimo_registro}")
            data_inicio = ultimo_registro + timedelta(days=1)
            primeira_extracao = False

    print(f"Extraindo dados de {data_inicio} até {data_fim}...")

    try:
        df_novo = yf.download(
            ticker,
            start=data_inicio.strftime("%Y-%m-%d"),
            end=(data_fim + timedelta(days=1)).strftime("%Y-%m-%d"),
            interval=intervalo,
            progress=False,
        )
    except Exception as e:
        print(f"Erro ao extrair dados do Yahoo Finance: {e}")
        return df_existente if not df_existente.empty else pd.DataFrame()

    print(f"\nDados extraídos: {len(df_novo)} registros.")

    if df_novo.empty:
        print("\nNenhum dado complementar foi adicionado.")
        return df_existente if not df_existente.empty else pd.DataFrame()

    # Garantir index e colunas corretos
    df_novo.index = pd.to_datetime(df_novo.index, utc=True).tz_convert('America/Sao_Paulo').tz_localize(None)
    df_novo.columns = df_novo.columns.to_flat_index()
    df_novo.columns = [col[0] if isinstance(col, tuple) else col for col in df_novo.columns]

    if primeira_extracao:
        primeira_data = df_novo.index[0].normalize()
        df_novo = df_novo[df_novo.index.normalize() != primeira_data]
        print("\nAmostra após remoção da primeira data:")
        df_novo = df_novo.sort_index()
        print(df_novo.head(3))

    df_novo_limpo = limpeza_dados(df_novo)

    if primeira_extracao:
        print("\nPrimeira extração de dados.")
        df_total_limpo = df_novo_limpo.copy()
        os.makedirs(os.path.dirname(path_dados_limpos), exist_ok=True)
        os.makedirs(os.path.dirname(path_dados_brutos), exist_ok=True)
        df_total_limpo.to_csv(path_dados_brutos, index=False)
        df_total_limpo.to_csv(path_dados_limpos, index=False)
    else:
        print("\nExtração complementar realizada.")
        df_existente = df_existente[df_novo_limpo.columns]  # garantir colunas consistentes
        df_total_limpo = pd.concat([df_existente, df_novo_limpo])
        df_total_limpo.drop_duplicates(subset=["datetime"], keep="first", inplace=True)
        df_total_limpo = df_total_limpo.sort_values(["data", "hora"])
        df_total_limpo.to_csv(path_dados_limpos, index=False)

    print("\nExtração e limpeza finalizadas com sucesso.")
    return df_total_limpo

# Execução direta (testes)
if __name__ == "__main__":
    ticker = 'BBDC4.SA'
    intervalo = '5m'
    path_projeto = "/content/Piloto_day_trade"
    df_total_limpo = extrair_dados(ticker, intervalo, path_projeto)
```

```python
%%writefile /content/Piloto_day_trade/src/operacional/diagnostico_dados_iniciais.py

#@title Script Diagnóstico de consistência dos dados limpos

import pandas as pd
import os
from datetime import datetime

def diagnostico_dados_iniciais(df: pd.DataFrame, n=20, salvar_em_txt=True):
    """
    Executa diagnósticos sobre os primeiros registros do dataset.

    Parâmetros:
        df (pd.DataFrame): DataFrame com colunas ['data', 'hora', 'abertura', 'minimo', 'maximo', 'fechamento', 'volume', 'datetime']
        n (int): Quantidade de registros iniciais a avaliar
        salvar_em_txt (bool): Se True, salva relatório em reports/
    """
    relatorio = []

    relatorio.append("=== Diagnóstico Inicial dos Dados ===\n")

    # 1. Informações gerais
    relatorio.append(f"Total de registros: {len(df)}")
    relatorio.append(f"Período da série: {df['datetime'].min()} até {df['datetime'].max()}")
    relatorio.append(f"Frequência estimada: {df['datetime'].diff().mode()[0]}")

    # 2. Checagem de valores ausentes
    relatorio.append("\nValores nulos nos primeiros registros:")
    relatorio.append(str(df.head(n).isnull().sum()))

    # 3. Verificação de sequência temporal
    relatorio.append("\nIntervalos de tempo nos primeiros registros:")
    relatorio.append(str(df['datetime'].head(n).diff().value_counts(dropna=False)))

    # 4. Estatísticas de preços
    primeiros = df.head(n)
    relatorio.append("\nEstatísticas dos preços nos primeiros registros:")
    relatorio.append(str(primeiros[['abertura', 'minimo', 'maximo', 'fechamento']].describe()))

    relatorio.append("\nVolumes nos primeiros registros:")
    relatorio.append(str(primeiros['volume'].describe()))

    # Volumes zerados
    volumes_zerados = primeiros[primeiros['volume'] == 0]
    if not volumes_zerados.empty:
        relatorio.append(f"\nAtenção: {len(volumes_zerados)} registros com volume 0 nos primeiros {n} registros:")
        relatorio.append(str(volumes_zerados[['datetime', 'volume']]))

    # Inconsistência entre colunas
    if not pd.to_datetime(df['data'] + ' ' + df['hora']).equals(df['datetime']):
        relatorio.append("\nInconsistência entre 'data' + 'hora' e coluna 'datetime'. Verifique geração dos dados.")

    relatorio.append("\n=== Fim do Diagnóstico ===")

    # Impressão no console
    print("\n".join(relatorio))

    # Salvar relatório em arquivo txt
    if salvar_em_txt:
        os.makedirs("reports", exist_ok=True)
        data_execucao = datetime.now().strftime("%Y%m%d_%H%M%S")
        caminho_arquivo = f"data/cleaned/diagnostico_inicial_{data_execucao}.txt"
        with open(caminho_arquivo, "w", encoding="utf-8") as f:
            f.write("\n".join(relatorio))

if __name__ == "__main__":
    df_limpo= pd.read_csv('/content/Piloto_day_trade/data/transformed/dados_transformados_BBDC4.csv', parse_dates=['datetime'])
    diagnostico_dados_iniciais(df_limpo)
```

```python
%%writefile /content/Piloto_day_trade/src/pipeline_etl/transformacao_dados_versao001.py
#@title Script de transformação de dados

import os
import pandas as pd
import numpy as np
from ta.momentum import RSIIndicator
from ta.trend import EMAIndicator
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

def carregar_dados_limpos(path_dados_limpos: str) -> pd.DataFrame:
    print(f"Carregando dados limpos de: {path_dados_limpos}...")
    try:
        df = pd.read_csv(path_dados_limpos)
        df['datetime'] = pd.to_datetime(df['datetime'], errors='coerce')
        df.dropna(subset=['datetime'], inplace=True)
        df['data'] = pd.to_datetime(df['datetime'].dt.date)
        df['hora'] = df['datetime'].dt.time
        print(f"Dados limpos carregados. Total de {len(df)} registros.")
        return df
    except FileNotFoundError:
        print(f"[ERRO] Arquivo não encontrado: {path_dados_limpos}")
        return pd.DataFrame()

def carregar_dados_transformados_existentes(path_arquivo_transformado: str, df_limpo: pd.DataFrame) -> pd.DataFrame:
    if os.path.exists(path_arquivo_transformado):
        df_transformado_existente = pd.read_csv(path_arquivo_transformado, parse_dates=['datetime', 'data'])
        df_transformado_existente['data'] = pd.to_datetime(df_transformado_existente['data'])

        ultima_data_transformada = df_transformado_existente['data'].max().date()
        print(f"[INFO] Última data transformada existente: {ultima_data_transformada}")

        df_novas_datas_a_processar = df_limpo[df_limpo['data'].dt.date > ultima_data_transformada].copy()
        df_ultima_data_limpo = df_limpo[df_limpo['data'].dt.date == ultima_data_transformada].copy()
        df_ultima_data_existente = df_transformado_existente[df_transformado_existente['data'].dt.date == ultima_data_transformada].copy()

        novos_candles_ultima_data = df_ultima_data_limpo[
            ~df_ultima_data_limpo['datetime'].isin(df_ultima_data_existente['datetime'])
        ].copy()

        df = pd.concat([df_novas_datas_a_processar, novos_candles_ultima_data], ignore_index=True)
        df = df.sort_values('datetime').reset_index(drop=True)

        if df.empty:
            print("[AVISO] Nenhuma nova data ou novo candle disponível para transformação. Pipeline encerrado.")
            return pd.DataFrame()

        print(f"[INFO] Novas datas/candles a serem processadas: {df['data'].min().date()} a {df['data'].max().date()}. Total de {len(df)} registros.")
        return df
    else:
        print("[INFO] Nenhum arquivo transformado anterior encontrado. Serão processadas todas as datas disponíveis.")
        return df_limpo.copy()

def filtrar_janela_horaria(df: pd.DataFrame) -> pd.DataFrame:
    """
    Filtra o DataFrame para incluir apenas registros dentro do intervalo de negociação.
    """
    print("Filtrando registros fora do intervalo 10:05 - 16:50...")
    hora_inicio = pd.to_datetime("10:05").time()
    hora_fim = pd.to_datetime("16:50").time()
    df_filtrado = df[(df['hora'] >= hora_inicio) & (df['hora'] <= hora_fim)].copy()
    print(f"Total de {len(df_filtrado)} registros após filtragem horária.")
    return df_filtrado

def preencher_lacunas(df: pd.DataFrame) -> pd.DataFrame:
    """
    Preenche lacunas de tempo (candles ausentes de 5 em 5 minutos) usando a lógica de cópia do vizinho mais próximo.
    """
    print("Preenchendo lacunas de horário...")

    horarios_esperados_times = pd.date_range("10:05", "16:50", freq="5min").time

    linhas_adicionadas = []

    for data_unica_date_obj in df['data'].dt.date.unique():
        data_unica = data_unica_date_obj
        df_dia = df[df['data'].dt.date == data_unica].copy()

        full_datetimes_esperados = [pd.Timestamp.combine(data_unica, h) for h in horarios_esperados_times]

        df_full_day = pd.DataFrame({'datetime': full_datetimes_esperados})

        df_dia_merged = pd.merge(df_full_day, df_dia, on='datetime', how='left')

        horarios_faltantes_df = df_dia_merged[df_dia_merged['fechamento_candle'].isna()]

        for index, row in horarios_faltantes_df.iterrows():
            dt_faltante = row['datetime']

            anteriores = df_dia[df_dia['datetime'] < dt_faltante]
            posteriores = df_dia[df_dia['datetime'] > dt_faltante]

            copia_base = None

            if not anteriores.empty and not posteriores.empty:
                anterior = anteriores.iloc[-1]
                posterior = posteriores.iloc[0]

                if abs((dt_faltante - anterior['datetime']).total_seconds()) <= abs((posterior['datetime'] - dt_faltante).total_seconds()):
                    copia_base = anterior
                else:
                    copia_base = posterior
            elif not anteriores.empty:
                copia_base = anteriores.iloc[-1]
            elif not posteriores.empty:
                copia_base = posteriores.iloc[0]
            else:
                continue

            if copia_base is not None:
                nova_linha = copia_base.copy()
                nova_linha['datetime'] = dt_faltante
                nova_linha['data'] = pd.to_datetime(dt_faltante.date())
                nova_linha['hora'] = dt_faltante.time()
                nova_linha['volume_candle'] = 0
                nova_linha['ajuste'] = 0
                linhas_adicionadas.append(nova_linha)

    if linhas_adicionadas:
        df_adicional = pd.DataFrame(linhas_adicionadas)
        df_adicional['data'] = pd.to_datetime(df_adicional['data'])
        df = pd.concat([df, df_adicional], ignore_index=True)
        df = df.sort_values('datetime').reset_index(drop=True)
        print(f"Adicionadas {len(linhas_adicionadas)} lacunas. Total de {len(df)} registros.")
    else:
        print("Nenhuma lacuna encontrada ou preenchida.")
    return df

def carregar_historico_transformado_necessario(df_novas_datas: pd.DataFrame, path_dados_transformados: str, janela_minima: int = 20) -> pd.DataFrame:
    """
    Carrega um histórico recente de dados transformados para garantir o cálculo correto de indicadores.
    Retorna um DataFrame que inclui o histórico e as novas datas.
    """
    print("[INFO] Carregando histórico transformado para cálculo dos indicadores...")
    try:
        df_historico_transformado = pd.read_csv(path_dados_transformados, parse_dates=['datetime', 'data'])
    except FileNotFoundError:
        print("[AVISO] Arquivo de histórico transformado não encontrado. Prosseguindo apenas com as novas datas.")
        return df_novas_datas

    min_datetime_novas_datas = df_novas_datas['datetime'].min()

    dados_anteriores = df_historico_transformado[df_historico_transformado['datetime'] < min_datetime_novas_datas].copy()

    dados_suficientes = dados_anteriores.tail(janela_minima + 50)

    df_combinado = pd.concat([dados_suficientes, df_novas_datas], ignore_index=True).sort_values('datetime').reset_index(drop=True)

    print(f"Carregados {len(dados_suficientes)} registros históricos para suporte de cálculo.")
    return df_combinado

def calcular_indicadores(df: pd.DataFrame) -> pd.DataFrame:
    """
    Calcula os indicadores técnicos RSI e EMA, e trata NaNs resultantes.
    """
    print("Calculando indicadores técnicos (RSI, EMA)...")
    df['fechamento_candle'] = pd.to_numeric(df['fechamento_candle'], errors='coerce')

    df['RSI'] = RSIIndicator(close=df['fechamento_candle'], window=14).rsi()
    df['EMA_20'] = EMAIndicator(close=df['fechamento_candle'], window=20).ema_indicator()

    df['RSI_indicador_faltante'] = df['RSI'].isna().astype(int)
    df['EMA_20_indicador_faltante'] = df['EMA_20'].isna().astype(int)

    df['RSI'] = df['RSI'].interpolate(method='linear', limit_direction='both').fillna(0)
    df['EMA_20'] = df['EMA_20'].interpolate(method='linear', limit_direction='both').fillna(0)

    print("Cálculo de indicadores concluído.")
    return df

def adicionar_features_temporais(df: pd.DataFrame) -> pd.DataFrame:
    """
    Adiciona features baseadas em tempo ao DataFrame.
    """
    print("Adicionando features temporais...")
    df['hora_num'] = df['datetime'].dt.hour
    df['minuto'] = df['datetime'].dt.minute
    df['dia_semana'] = df['datetime'].dt.dayofweek
    print("Features temporais adicionadas.")
    return df

def calcular_retorno(df: pd.DataFrame) -> pd.DataFrame:
    """
    Calcula o retorno percentual de cada candle.
    """
    print("Calculando retorno percentual por candle...")
    df['retorno'] = df['fechamento_candle'].pct_change()
    df['retorno'].fillna(0, inplace=True)
    print("Retorno percentual calculado.")
    return df

def calcular_agregados_diarios(df: pd.DataFrame) -> pd.DataFrame:
    """
    Calcula os valores de abertura, mínima, máxima, fechamento e volume diários.
    """
    print("Calculando agregados diários...")
    df['abertura_dia'] = df.groupby('data')['abertura_candle'].transform('first')
    df['minima_dia'] = df.groupby('data')['min_candle'].transform('min')
    df['maxima_dia'] = df.groupby('data')['max_candle'].transform('max')
    df['fechamento_dia'] = df.groupby('data')['fechamento_candle'].transform('last')
    df['volume_dia'] = df.groupby('data')['volume_candle'].transform('sum')
    print("Agregados diários calculados.")
    return df

def criar_coluna_binaria_outliers(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cria uma coluna binária indicando a presença de outliers em colunas numéricas.
    """
    print("Criando coluna binária para outliers...")
    exclude_cols = ['is_outlier', 'RSI_indicador_faltante', 'EMA_20_indicador_faltante', 'hora_num', 'minuto', 'dia_semana']
    colunas_numericas = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
    colunas_para_outlier = [col for col in colunas_numericas if col not in exclude_cols]

    outlier_mask = pd.Series(False, index=df.index)

    for col in colunas_para_outlier:
        if df[col].isnull().sum() > len(df) * 0.5:
            continue

        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1
        limite_inferior = q1 - 1.5 * iqr
        limite_superior = q3 + 1.5 * iqr

        outliers_col = (df[col] < limite_inferior) | (df[col] > limite_superior)
        outlier_mask = outlier_mask | outliers_col

    df['is_outlier'] = outlier_mask.astype(int)
    print("Coluna 'is_outlier' criada.")
    return df

def salvar_ultimos_pregoes(df: pd.DataFrame, ticker: str, path_dados_transformados: str, n_pregoes: int = 3) -> None:
    """
    Salva os dados dos últimos n pregões em formato CSV para uso no modelo.
    """
    df['data'] = pd.to_datetime(df['data'])
    datas_recentes = sorted(df['data'].dt.date.unique())[-n_pregoes:]
    df_daily = df[df['data'].dt.date.isin(datas_recentes)].copy()

    pasta_saida = os.path.join(path_dados_transformados,'daily')
    os.makedirs(pasta_saida, exist_ok=True)
    caminho_saida = os.path.join(pasta_saida, f"dados_{n_pregoes}_ultimos_pregoes.csv")

    df_daily.to_csv(caminho_saida, index=False)
    print(f"[✓] Últimos {n_pregoes} pregões salvos em: {caminho_saida}")


def transformar_dados(ticker: str, path_projeto: str) -> None:
    print(f"\n--- Transformação de dados  iniciada ---\n")

    path_dados_limpos = os.path.join(path_projeto, f'data/cleaned/{ticker.split(".")[0]}/dados_limpos_{ticker.split(".")[0]}.csv')
    path_dados_transformados = os.path.join(path_projeto, f'data/transformed/{ticker.split(".")[0]}')
    path_arquivo_transformado = os.path.join(path_dados_transformados, f'dados_transformados_{ticker.split(".")[0]}.csv')

    df_limpo = carregar_dados_limpos(path_dados_limpos)
    if df_limpo.empty:
        print("[AVISO] Nenhum dado carregado. Encerrando transformação.")
        return

    df_limpo.rename(columns={
        'abertura': 'abertura_candle',
        'minimo': 'min_candle',
        'maximo': 'max_candle',
        'fechamento': 'fechamento_candle',
        'volume': 'volume_candle'
    }, inplace=True)

    df = carregar_dados_transformados_existentes(path_arquivo_transformado, df_limpo)
    if df.empty:
        return

    df = filtrar_janela_horaria(df)
    df = preencher_lacunas(df)

    if os.path.exists(path_arquivo_transformado):
        df_para_indicadores = carregar_historico_transformado_necessario(df, path_arquivo_transformado)
    else:
        df_para_indicadores = df.copy()

    df_para_indicadores = calcular_indicadores(df_para_indicadores)
    df_para_indicadores = adicionar_features_temporais(df_para_indicadores)
    df_para_indicadores = calcular_retorno(df_para_indicadores)
    df_para_indicadores = calcular_agregados_diarios(df_para_indicadores)
    df_para_indicadores = criar_coluna_binaria_outliers(df_para_indicadores)

    if os.path.exists(path_arquivo_transformado):
        df_final = df_para_indicadores[df_para_indicadores['datetime'] >= df['datetime'].min()].copy()
    else:
        df_final = df_para_indicadores.copy()

    colunas_finais = [
        'data', 'hora',
        'abertura_dia', 'minima_dia', 'maxima_dia', 'fechamento_dia', 'volume_dia',
        'abertura_candle', 'min_candle', 'max_candle', 'fechamento_candle', 'volume_candle',
        'RSI', 'EMA_20', 'retorno',
        'dia_semana', 'hora_num', 'minuto', 'datetime', 'is_outlier',
        'RSI_indicador_faltante', 'EMA_20_indicador_faltante'
    ]

    for col in colunas_finais:
        if col not in df_final.columns:
            df_final[col] = np.nan

    df_final = df_final[colunas_finais].copy()
    df_final.fillna(0, inplace=True)

    if os.path.exists(path_arquivo_transformado):
        df_existente_para_anexar = pd.read_csv(path_arquivo_transformado, parse_dates=['datetime', 'data'])
        df_existente_para_anexar['data'] = pd.to_datetime(df_existente_para_anexar['data'])
        df_salvar = pd.concat([df_existente_para_anexar, df_final], ignore_index=True)
        df_salvar.drop_duplicates(subset=['datetime'], inplace=True)
        df_salvar.sort_values('datetime', inplace=True)
        df_salvar.reset_index(drop=True, inplace=True)
    else:
        df_salvar = df_final.copy()

    os.makedirs(os.path.dirname(path_arquivo_transformado), exist_ok=True)
    df_salvar.to_csv(path_arquivo_transformado,  index=False)
    print(f"\n[✓] Transformação concluída. Dados salvos em: {path_arquivo_transformado}. Total de {len(df_salvar)} registros.")

    salvar_ultimos_pregoes(df_salvar, ticker, path_projeto)

if __name__ == "__main__":
    ticker = 'BBDC4.SA'
    path_projeto = "/content/Piloto_day_trade"
    transformar_dados(ticker, path_projeto)

```

```python
#@title Script diagnotico para os dados transformados
%%writefile /content/Piloto_day_trade/src/operacional/diagnostico_dados_transformados.py

import pandas as pd
import numpy as np
import os
import glob

def diagnostico_dados(df: pd.DataFrame, pasta_relatorios: str = "reports") -> dict:
    resultados = {}
    data_base = df['datetime'].dt.date.max().strftime("%Y-%m-%d")
    nome_arquivo = f"diagnostico_consistencia_dados_{data_base}.txt"
    caminho_pasta = os.path.join(os.getcwd(), pasta_relatorios)
    caminho_arquivo = os.path.join(caminho_pasta, nome_arquivo)

    if not os.path.exists(caminho_pasta):
        os.makedirs(caminho_pasta)

    linhas = []
    linhas.append(f"Relatório de Diagnóstico - Dados de {data_base}\n")
    linhas.append("Verificações de Consistência:\n")

    # 1. Ausência de nulos
    nulos = df.isnull().sum()
    for col, val in nulos.items():
        if val > 0:
            linhas.append(f"Aviso: {val} valores nulos em '{col}'.\n")
        else:
            linhas.append(f"OK: Coluna '{col}' sem nulos.\n")
    resultados['nulos'] = nulos.to_dict()

    # 2. Consistência de volume
    volume = df.get('volume')
    if volume is not None:
        negativos = (volume < 0).sum()
        zeros = (volume == 0).sum()
        if negativos > 0:
            linhas.append(f"Erro: {negativos} valores negativos em volume.\n")
        else:
            linhas.append("OK: Nenhum volume negativo.\n")
        if zeros > 0:
            linhas.append(f"Aviso: {zeros} volumes zerados.\n")
        else:
            linhas.append("OK: Nenhum volume zero.\n")
        resultados['volume_negativo'] = negativos
        resultados['volume_zero'] = zeros

    # 3. Tipos corretos
    tipos_incorretos = {
        col: str(dtype)
        for col, dtype in df.dtypes.items()
        if not (pd.api.types.is_numeric_dtype(dtype) or pd.api.types.is_datetime64_any_dtype(dtype))
    }
    if tipos_incorretos:
        for col, tipo in tipos_incorretos.items():
            linhas.append(f"Erro: Tipo incorreto na coluna '{col}' ({tipo}).\n")
    else:
        linhas.append("OK: Todos os tipos de dados estão corretos.\n")
    resultados['tipos_incorretos'] = tipos_incorretos

    # 4. Ausência de duplicatas
    duplicatas = df.duplicated().sum()
    if duplicatas > 0:
        linhas.append(f"Aviso: {duplicatas} linhas duplicadas.\n")
    else:
        linhas.append("OK: Nenhuma duplicata.\n")
    resultados['duplicatas'] = duplicatas

    # 5. Consistência temporal
    if 'datetime' in df.columns:
        df_sorted = df.sort_values('datetime')
        ordem_ok = df_sorted['datetime'].is_monotonic_increasing
        linhas.append("OK: Ordem temporal correta.\n" if ordem_ok else "Erro: Dados fora de ordem temporal.\n")
        resultados['ordem_temporal'] = ordem_ok

        diffs = df_sorted['datetime'].diff().value_counts().to_dict()
        resultados['intervalos_temporais'] = {str(k): int(v) for k, v in diffs.items()}
    else:
        linhas.append("Erro: Coluna 'datetime' ausente.\n")
        resultados['ordem_temporal'] = False
        resultados['intervalos_temporais'] = {}

    # 6. Outliers extremos (z-score > 5)
    df_numeric = df.select_dtypes(include=[np.number])
    zscores = (df_numeric - df_numeric.mean()) / df_numeric.std()
    outliers = (np.abs(zscores) > 5).sum()
    for col, qtde in outliers.items():
        if qtde > 0:
            linhas.append(f"Aviso: {qtde} outliers extremos em '{col}'.\n")
    resultados['outliers'] = outliers.to_dict()

    # 7. Checagem dos indicadores
    indicadores_check = {}
    fechamento = df.get('fechamento')
    if fechamento is not None:
        if 'SMA_10' in df.columns:
            sma_teorico = fechamento.rolling(10).mean()
            match_pct = np.isclose(df['SMA_10'], sma_teorico, rtol=1e-3, atol=1e-4).mean()
            indicadores_check['SMA_10'] = match_pct
            linhas.append(f"SMA_10: {match_pct:.2%} dos valores coerentes.\n")
        if 'EMA_10' in df.columns:
            ema_teorico = fechamento.ewm(span=10, adjust=False).mean()
            match_pct = np.isclose(df['EMA_10'], ema_teorico, rtol=1e-3, atol=1e-4).mean()
            indicadores_check['EMA_10'] = match_pct
            linhas.append(f"EMA_10: {match_pct:.2%} dos valores coerentes.\n")
    resultados['checagem_indicadores'] = indicadores_check

    # 8. Verificação de coerência entre features derivadas
    coerencia = {}
    if 'retorno_log' in df.columns and fechamento is not None:
        retorno_teorico = np.log(fechamento / fechamento.shift(1))
        match_pct = np.isclose(df['retorno_log'], retorno_teorico, rtol=1e-4, atol=1e-6).mean()
        coerencia['retorno_log'] = match_pct
        linhas.append(f"retorno_log: {match_pct:.2%} dos valores coerentes.\n")
    if 'volatilidade' in df.columns and 'retorno_log' in df.columns:
        vol_teorica = df['retorno_log'].rolling(window=10).std()
        match_pct = np.isclose(df['volatilidade'], vol_teorica, rtol=1e-3, atol=1e-6).mean()
        coerencia['volatilidade'] = match_pct
        linhas.append(f"volatilidade: {match_pct:.2%} dos valores coerentes.\n")
    resultados['coerencia_features'] = coerencia

    # Impressão em tela
    print("".join(linhas))

    # Salvamento do relatório
    with open(caminho_arquivo, 'w', encoding='utf-8') as f:
        f.writelines(linhas)

    # Gerenciamento: manter apenas os 10 arquivos mais recentes
    arquivos = sorted(glob.glob(os.path.join(caminho_pasta, "diagnostico_consistencia_dados_*.txt")),
                      key=os.path.getmtime)
    while len(arquivos) > 10:
        os.remove(arquivos[0])
        arquivos.pop(0)

    return resultados

if __name__ == "__main__":
    path_dados_transformados = '/content/Piloto_day_trade/data/transformed/dados_transformados_BBDC4.csv'
    df = pd.read_csv(path_dados_transformados, parse_dates=['datetime'])
    diagnostico_dados(df)

```

```python
#@title Script de carga de dados em banco SQLite
%%writefile /content/Piloto_day_trade/src/pipeline_etl/carga_dados.py
import sqlite3
import pandas as pd
from datetime import datetime
import os

def conectar_db(caminho_db):
    """Conecta ou cria o banco de dados SQLite."""
    return sqlite3.connect(caminho_db)

def criar_tabelas(conn):
    """Cria as tabelas no banco de dados se elas não existirem."""
    cursor = conn.cursor()

    # Tabela de dimensão de tempo
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS dim_tempo (
        id_tempo INTEGER PRIMARY KEY AUTOINCREMENT,
        data TEXT NOT NULL,
        hora TEXT NOT NULL,
        datetime TEXT NOT NULL UNIQUE, -- datetime é a chave de unicidade para o ponto no tempo
        dia_da_semana INTEGER NOT NULL
    );
    """)

    # Tabela de dimensão de indicadores técnicos
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS dim_indicadores (
        id_indicadores INTEGER PRIMARY KEY AUTOINCREMENT,
        RSI REAL NOT NULL,
        EMA_20 REAL NOT NULL,
        retorno REAL NOT NULL,
        UNIQUE(RSI, EMA_20, retorno) -- Garante unicidade dos valores dos indicadores
    );
    """)

    # Tabela de dimensão de agregados diários
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS dim_agregados (
        id_agregados INTEGER PRIMARY KEY AUTOINCREMENT,
        hora_num INTEGER NOT NULL,
        minuto INTEGER NOT NULL,
        abertura_dia REAL NOT NULL,
        minima_dia REAL NOT NULL,
        maxima_dia REAL NOT NULL,
        fechamento_dia REAL NOT NULL,
        volume_dia INTEGER NOT NULL,
        UNIQUE(hora_num, minuto, abertura_dia, minima_dia, maxima_dia, fechamento_dia, volume_dia)
    );
    """)

    # Tabela de fatos para os preços e suas relações com as dimensões
    cursor.execute("""
    CREATE TABLE IF NOT EXISTS fato_precos (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        abertura REAL NOT NULL,
        minima REAL NOT NULL,
        maxima REAL NOT NULL,
        fechamento REAL REAL NOT NULL,
        volume INTEGER NOT NULL,
        id_tempo INTEGER NOT NULL,
        id_indicadores INTEGER NOT NULL,
        id_agregados INTEGER NOT NULL,
        FOREIGN KEY (id_tempo) REFERENCES dim_tempo(id_tempo),
        FOREIGN KEY (id_indicadores) REFERENCES dim_indicadores(id_indicadores),
        FOREIGN KEY (id_agregados) REFERENCES dim_agregados(id_agregados),
        UNIQUE(id_tempo, id_indicadores, id_agregados) -- Garante unicidade da linha de fato
    );
    """)

    conn.commit()

def obter_id(conn, tabela, valores):
    """
    Obtém o ID de um registro existente na dimensão ou insere um novo, retornando o ID.
    Utiliza INSERT OR IGNORE para otimizar o processo e garantir a unicidade dos dados.
    """
    placeholders = ','.join(['?'] * len(valores))
    colunas = ','.join(valores.keys())
    cursor = conn.cursor()

    # Tenta inserir o registro. Se já existir (violando a constraint UNIQUE), a inserção é ignorada.
    cursor.execute(f"INSERT OR IGNORE INTO {tabela} ({colunas}) VALUES ({placeholders})", tuple(valores.values()))

    # Seleciona o ID (rowid) do registro, seja ele recém-inserido ou já existente.
    query = f"SELECT rowid FROM {tabela} WHERE " + ' AND '.join([f"{k}=?" for k in valores])
    cursor.execute(query, tuple(valores.values()))
    row = cursor.fetchone()

    return row[0] # Retorna o ID do registro

def carregar_dados(ticker: str, path_projeto: str) -> None:
    """
    Carrega dados transformados de um arquivo CSV para um banco de dados SQLite,
    otimizando a inserção e gerenciando as tabelas de dimensões e fatos.
    """
    print('\nIniciando a carga de dados...')

    # Define os caminhos para o arquivo CSV de entrada e o banco de dados SQLite de saída
    caminho_csv = os.path.join(path_projeto, f'data/transformed/{ticker.split(".")[0]}/dados_transformados_{ticker.split(".")[0]}.csv')
    caminho_db = os.path.join(path_projeto, f'db/{ticker.split(".")[0]}/SQLite_{ticker.split(".")[0]}.db')

    # Cria o diretório 'db' se ele não existir
    os.makedirs(os.path.dirname(caminho_db), exist_ok=True)

    # Verifica se o arquivo CSV de entrada existe
    if not os.path.exists(caminho_csv):
        print(f"[ERRO] Arquivo CSV não encontrado: {caminho_csv}")
        return

    # Carrega os dados do CSV para um DataFrame, parseando a coluna 'datetime' como tipo data/hora
    df = pd.read_csv(caminho_csv, parse_dates=['datetime'])

    # Verifica se o DataFrame carregado está vazio
    if df.empty:
        print("[INFO] Nenhum dado para inserir.")
        return

    # Conecta-se ao banco de dados e cria as tabelas se necessário
    conn = conectar_db(caminho_db)
    criar_tabelas(conn)

    print("Processando e inserindo dados nas tabelas de dimensão...")

    # Dicionários de cache para armazenar IDs de dimensões já processadas e evitar consultas repetidas
    cache_dim_tempo = {}
    cache_dim_indicadores = {}
    cache_dim_agregados = {}

    # Lista para coletar os valores das linhas da tabela de fatos para inserção em batch
    fato_valores_para_inserir = []

    # Itera sobre cada linha do DataFrame para processar e coletar dados
    for index, row in df.iterrows():
        # Formata as colunas 'data', 'hora' e 'datetime' como strings para inserção no SQLite
        data_str = row['data']
        hora_str = row['hora']
        datetime_str = row['datetime'].isoformat() # Converte Timestamp para string ISO 8601

        # Processa e obtém o ID para a dimensão de tempo
        valores_tempo_tuple = (data_str, hora_str, datetime_str, row['dia_semana'])
        if valores_tempo_tuple not in cache_dim_tempo:
            id_tempo = obter_id(conn, 'dim_tempo', {
                'data': data_str,
                'hora': hora_str,
                'datetime': datetime_str,
                'dia_da_semana': row['dia_semana']
            })
            cache_dim_tempo[valores_tempo_tuple] = id_tempo
        else:
            id_tempo = cache_dim_tempo[valores_tempo_tuple]

        # Processa e obtém o ID para a dimensão de indicadores
        valores_ind_tuple = (row['RSI'], row['EMA_20'], row['retorno'])
        if valores_ind_tuple not in cache_dim_indicadores:
            id_ind = obter_id(conn, 'dim_indicadores', {
                'RSI': row['RSI'], 'EMA_20': row['EMA_20'], 'retorno': row['retorno']
            })
            cache_dim_indicadores[valores_ind_tuple] = id_ind
        else:
            id_ind = cache_dim_indicadores[valores_ind_tuple]

        # Processa e obtém o ID para a dimensão de agregados
        valores_agg_tuple = (
            row['hora_num'], row['minuto'],
            row['abertura_dia'], row['minima_dia'], row['maxima_dia'],
            row['fechamento_dia'], row['volume_dia']
        )
        if valores_agg_tuple not in cache_dim_agregados:
            id_agg = obter_id(conn, 'dim_agregados', {
                'hora_num': row['hora_num'], 'minuto': row['minuto'],
                'abertura_dia': row['abertura_dia'],
                'minima_dia': row['minima_dia'],
                'maxima_dia': row['maxima_dia'],
                'fechamento_dia': row['fechamento_dia'],
                'volume_dia': row['volume_dia']
            })
            cache_dim_agregados[valores_agg_tuple] = id_agg
        else:
            id_agg = cache_dim_agregados[valores_agg_tuple]

        # Adiciona os valores da linha de fato à lista para inserção em batch
        fato_valores_para_inserir.append((
            row['abertura_candle'], row['min_candle'], row['max_candle'], row['fechamento_candle'],
            row['volume_candle'],
            id_tempo, id_ind, id_agg
        ))

    # Realiza a inserção em batch para a tabela de fatos
    cursor = conn.cursor()
    try:
        cursor.executemany("""
            INSERT OR IGNORE INTO fato_precos (
                abertura, minima, maxima, fechamento, volume,
                id_tempo, id_indicadores, id_agregados
            ) VALUES (?, ?, ?, ?, ?, ?, ?, ?)
        """, fato_valores_para_inserir)
        conn.commit() # Confirma a transação
        novos_registros = cursor.rowcount # Obtém o número de linhas realmente inseridas
    except sqlite3.Error as e:
        conn.rollback() # Em caso de erro, desfaz a transação
        print(f"[ERRO] Erro durante a inserção de fato_precos: {e}")
        novos_registros = 0

    conn.close() # Fecha a conexão com o banco de dados
    print(f"Carga concluída com {novos_registros} novos registros na tabela fato_precos.")

# Execução direta poara testes
if __name__ == "__main__":
    carregar_dados('LREN3', '/content/Piloto_day_trade')
```

```python
# %%writefile /content/Piloto_day_trade/src/modelagem_ml/preparar_dados_lstm

#@title --- Preparar dados modelo LSTM ---

# Importa bibliotecas
import os
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import MinMaxScaler, StandardScaler


def normalizar_padronizar_dados(ticker, path_dados_transformados, path_scalers, path_dados_normalizados):
    df = pd.read_csv(path_dados_transformados)

    # Ordena corretamente por data e hora
    df = df.sort_values(by=['data', 'hora']).reset_index(drop=True)

    # Seleciona as colunas para o modelo
    df = df[['data', 'hora', 'dia_semana', 'abertura_candle', 'min_candle', 'max_candle',
             'fechamento_candle', 'volume_candle', 'RSI', 'EMA_20', 'retorno', 'hora_num', 'minuto']]

    # Aplica one-hot encoding na coluna de dia da semana para representar dias como variáveis binárias
    df = pd.get_dummies(df, columns=['dia_semana'], prefix='dia_semana')

    # Define colunas a serem normalizadas (MinMaxScaler) e padronizadas (StandardScaler)
    colunas_normalizar = ['abertura_candle', 'min_candle', 'max_candle', 'fechamento_candle', 'volume_candle']
    colunas_padronizar = ['RSI', 'EMA_20', 'retorno', 'hora_num', 'minuto']

    # Verifica se todas as colunas são numéricas
    for col in colunas_normalizar + colunas_padronizar:
        if not pd.api.types.is_numeric_dtype(df[col]):
            raise ValueError(f"Coluna {col} não é numérica.")

    # Aplica normalização nas colunas definidas
    minmax_scaler = MinMaxScaler()
    df[colunas_normalizar] = minmax_scaler.fit_transform(df[colunas_normalizar])

    # Aplica padronização nas colunas restantes
    standard_scaler = StandardScaler()
    df[colunas_padronizar] = standard_scaler.fit_transform(df[colunas_padronizar])

    # Salva os scalers treinados
    os.makedirs(path_scalers, exist_ok=True)
    joblib.dump(minmax_scaler, os.path.join(path_scalers, 'minmax_scaler.pkl'))
    joblib.dump(standard_scaler, os.path.join(path_scalers, 'standard_scaler.pkl'))

    scalers_info = {
        'colunas_normalizadas': colunas_normalizar,
        'colunas_padronizadas': colunas_padronizar,
        'colunas_one_hot': [col for col in df.columns if col.startswith('dia_semana_')]
    }
    joblib.dump(scalers_info, os.path.join(path_scalers, 'scalers_info.pkl'))

    # Salva os dados normalizados em CSV
    os.makedirs(path_dados_normalizados, exist_ok=True)
    df.to_csv(os.path.join(path_dados_normalizados, 'dados_normalizados.csv'), index=False)
    return df


# Função que cria as sequências de entrada (X) e alvo (y) para o modelo LSTM
# Considera dois dias como entrada e um dia como saída (completo de 82 candles)

def criar_sequencias(df, obs_dia, target_col, colunas_extra, path_sequencias):
    X, y, extras = [], [], []  # Listas para armazenar entradas, alvos e dados auxiliares
    seq_len = 2 * obs_dia  # Número total de observações de entrada (2 dias)
    n_alvos =  obs_dia   # Número de observações alvo (1 dia)

    # Loop para gerar todas as janelas válidas
    for i in range(len(df) - seq_len - n_alvos + 1):
        seq = df.iloc[i : i + seq_len]  # Entrada
        alvo = df.iloc[i + seq_len : i + seq_len + n_alvos]  # Saída

        datas_alvo = alvo['data'].unique()  # Verifica se o alvo pertence a um único pregão
        if len(datas_alvo) != 1:
            continue

        seq_x = seq.drop(columns=['data', 'hora'] + colunas_extra).values.astype(np.float32)  # Remove colunas extras
        seq_y = alvo[target_col].values.astype(np.float32)  # Extrai valores do alvo
        X.append(seq_x)
        y.append(seq_y)
        extras.append(alvo.iloc[0][colunas_extra].values)  # Armazena info extra do primeiro candle do alvo

    # Converte listas para arrays e salva em arquivos .npy
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.float32)
    extras = np.array(extras, dtype=object)

    os.makedirs(path_sequencias, exist_ok=True)
    np.save(os.path.join(path_sequencias, 'X.npy'), X)
    np.save(os.path.join(path_sequencias, 'y.npy'), y)
    np.save(os.path.join(path_sequencias, 'extras.npy'), extras)

    return X, y, extras

# Função que divide os dados em conjuntos de treino, validação e teste conforme proporções definidas
# E salva cada subconjunto em arquivos distintos para uso posterior

def dividir_dados(X, y, extras, path_treino_teste_val, proporcao_treino=0.7, proporcao_val=0.15):
    n = len(X)
    n_treino = int(n * proporcao_treino)
    n_val = int(n * proporcao_val)

    X_treino, y_treino, extras_treino = X[:n_treino], y[:n_treino], extras[:n_treino]
    X_val, y_val, extras_val = X[n_treino:n_treino + n_val], y[n_treino:n_treino + n_val], extras[n_treino:n_treino + n_val]
    X_teste, y_teste, extras_teste = X[n_treino + n_val:], y[n_treino + n_val:], extras[n_treino + n_val:]

    os.makedirs(path_treino_teste_val, exist_ok=True)
    np.save(os.path.join(path_treino_teste_val, 'X_treino.npy'), X_treino)
    np.save(os.path.join(path_treino_teste_val, 'y_treino.npy'), y_treino)
    np.save(os.path.join(path_treino_teste_val, 'extras_treino.npy'), extras_treino)

    np.save(os.path.join(path_treino_teste_val, 'X_val.npy'), X_val)
    np.save(os.path.join(path_treino_teste_val, 'y_val.npy'), y_val)
    np.save(os.path.join(path_treino_teste_val, 'extras_val.npy'), extras_val)

    np.save(os.path.join(path_treino_teste_val, 'X_teste.npy'), X_teste)
    np.save(os.path.join(path_treino_teste_val, 'y_teste.npy'), y_teste)
    np.save(os.path.join(path_treino_teste_val, 'extras_teste.npy'), extras_teste)

    return (X_treino, y_treino, extras_treino,
            X_val, y_val, extras_val,
            X_teste, y_teste, extras_teste)

# Função principal que define os caminhos e orquestra o processo de preparação dos dados
def preparar_dados_lstm(ticker, versao_dados, path_projeto):
    print(f"\n--- Preparando dados para o modelo LSTM para {ticker} na versão {versao_dados} ---")

    # Define os caminhos
    path_dados_transformados = os.path.join(path_projeto, f'data/transformed/{ticker}/dados_transformados_{ticker}.csv')
    base_dados = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}')
    path_dados_normalizados = os.path.join(base_dados, 'dados_normalizados')
    path_scalers = os.path.join(base_dados, 'scalers')
    path_sequencias = os.path.join(base_dados, 'sequencias')
    path_treino_teste_val = os.path.join(base_dados, 'treino_teste_val')

    # Executa a normalização
    print('\n[✔] Normalizando e padronizando dados...')
    df = normalizar_padronizar_dados(ticker, path_dados_transformados, path_scalers, path_dados_normalizados)

    print('\n[✔] Criando sequências de entrada e alvo...')
    obs_dia = 82  # Número de candles por dia (Total de observações)
    target_col = 'fechamento_candle'  # Coluna alvo para previsão
    colunas_extra = ['data', 'hora']  # Informações auxiliares que serão preservadas separadamente
    # Criação de sequências
    X, y, extras = criar_sequencias(df, obs_dia, target_col, colunas_extra, path_sequencias)
    # Divisão dos dados
    print('\n[✔] Dividindo dados em treino, validação e teste...')

    print(f"\nOs dados normnalizados estão em: {path_dados_normalizados}")
    print(f"Os scalers estão em: {path_scalers}")
    print(f"As sequências estão em: {path_sequencias}")
    print(f"Os conjuntos de treino, validação e teste estão em: {path_treino_teste_val}")

    return dividir_dados(X, y, extras, path_treino_teste_val)

# Execução direta (Usada para testes)
if __name__ == "__main__":
    preparar_dados_lstm('BBDC4', 'v001', '/content/Piloto_day_trade')
    print("\n[✔] Dados preparados com sucesso.")
```

```python
# %%writefile /content/Piloto_day_trade/src/modelagem_ml/treinar_lstm_v009.py
#@title --- Treinar modelo LSTM v009 ---

import os
import pickle
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import Huber

# from src.modelagem_ml.preparar_dados_lstm_v009 import preparar_dados_lstm
from src.modelagem_ml.avaliar_lstm import avaliar_lstm

import warnings
warnings.filterwarnings("ignore")

def treinar_lstm(ticker, versao_modelo, versao_dados, path_projeto, parametros: dict):
    print("Iniciando treinamento do modelo LSTM...")

    # Define os caminhos
    path_modelo_pasta = os.path.join(path_projeto, f'models/LSTM/{ticker}/modelo_lstm_{versao_modelo}')
    path_modelo = os.path.join(path_modelo_pasta, f'modelo_lstm_{versao_modelo}.h5')
    path_parametros_modelo = os.path.join(path_modelo_pasta, f'parametros_{versao_modelo}.pkl')
    path_treino_teste_val = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}/treino_teste_val')

    os.makedirs(path_modelo_pasta, exist_ok=True)

    # Carregar dados
    X_treino = np.load(os.path.join(path_treino_teste_val, 'X_treino.npy'), allow_pickle=True)
    y_treino = np.load(os.path.join(path_treino_teste_val, 'y_treino.npy'), allow_pickle=True)
    X_val = np.load(os.path.join(path_treino_teste_val, 'X_val.npy'), allow_pickle=True)
    y_val = np.load(os.path.join(path_treino_teste_val, 'y_val.npy'), allow_pickle=True)

    # Ajustar dimensões
    if len(X_treino.shape) == 2:
        X_treino = X_treino.reshape((X_treino.shape[0], X_treino.shape[1], 1))
        X_val = X_val.reshape((X_val.shape[0], X_val.shape[1], 1))

    # Verificação para garantir que tenhamos multiplas saidas
    if len(y_treino.shape) == 1:
        y_treino = y_treino.reshape((y_treino.shape[0], -1))
        y_val = y_val.reshape((y_val.shape[0], -1))

    # Configurar otimizador
    otimizador_nome = parametros['otimizador'].lower()
    taxa_aprendizado = parametros['taxa_aprendizado']
    if otimizador_nome == 'adam':
        otimizador = Adam(learning_rate=taxa_aprendizado)
    else:
        raise ValueError(f"Otimizador '{otimizador_nome}' não implementado.")

    # Configurar função perda
    perda_nome = parametros['funcao_perda'].lower()
    if perda_nome == 'huber':
        funcao_perda = Huber()
    else:
        raise ValueError(f"Função de perda '{perda_nome}' não implementada.")

    # Construção do modelo
    modelo = Sequential([
        LSTM(parametros['unidades_lstm_camada1'], return_sequences=True,
             input_shape=(X_treino.shape[1], X_treino.shape[2])),
        Dropout(parametros['taxa_dropout']),
        LSTM(parametros['unidades_lstm_camada2'], return_sequences=False),
        Dropout(parametros['taxa_dropout']),
        Dense(y_treino.shape[1])  # Saída com tamanho igual ao número de candles do pregão
    ])

    modelo.compile(optimizer=otimizador, loss=funcao_perda)

    # Callbacks
    callbacks = [
        EarlyStopping(monitor='val_loss', patience=parametros['paciencia_parada_antecipada'], restore_best_weights=True),
        ModelCheckpoint(path_modelo, monitor='val_loss', save_best_only=True)
    ]

    # Treinamento
    historico = modelo.fit(
        X_treino, y_treino,
        validation_data=(X_val, y_val),
        epochs=parametros['epocas_maximas'],
        batch_size=parametros['tamanho_lote'],
        callbacks=callbacks,
        verbose=1
    )

    print(f"[✓] Modelo treinado e salvo em: {path_modelo}")
    print(f"Total de épocas executadas: {len(historico.history['loss'])}")

    # Captura real dos parâmetros finais a partir do modelo
    config_modelo = modelo.get_config()
    parametros_modelo = {
        'ticker': ticker,
        'versao': versao_modelo,
        'path_modelo': path_modelo,
        'path_projeto': path_projeto,
        'epocas_executadas': len(historico.history['loss']),
        'loss_final_treino': historico.history['loss'][-1],
        'loss_final_val': historico.history['val_loss'][-1],
        'layers': [layer['class_name'] for layer in config_modelo['layers']]
    }

    print("[✓] Parâmetros finais do modelo:")
    for chave, valor in parametros_modelo.items():
        print(f"  {chave}: {valor}")

    with open(path_parametros_modelo, 'wb') as f:
        pickle.dump(parametros_modelo, f)

    print(f"[✓] Parâmetros de treinamento salvos em: {path_parametros_modelo}")

    return modelo, parametros_modelo

# Execução direta para testes
if __name__ == '__main__':
    parametros = {
        "dias_retrocesso": 2,
        "observacoes_por_dia": 82,
        "num_features_por_observacao": 20,
        "unidades_lstm_camada1": 128,
        "unidades_lstm_camada2": 128,
        "taxa_dropout": 0.25,
        "otimizador": "Adam",
        "taxa_aprendizado": 0.006,
        "funcao_perda": "Huber",
        "paciencia_parada_antecipada": 25,
        "epocas_maximas": 300,
        "tamanho_lote": 32
    }
    ticker = 'BBDC4'
    versao_modelo = 'v009'
    versao_dados = 'data_v009'
    path_projeto = '/content/Piloto_day_trade'

    # preparar_dados_lstm(ticker, versao_dados, path_projeto)
    treinar_lstm(ticker, versao_modelo, versao_dados, path_projeto, parametros)
    avaliar_lstm(ticker, versao_modelo, versao_dados, path_projeto)
```

```python
#@title Pipeline LSTM (Backup)

# Importa bibliotecas
import os
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import os
import pandas as pd
import numpy as np
import joblib
import json
import pickle
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from tensorflow.keras.models import load_model
import os
import pickle
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import Huber

import warnings
warnings.filterwarnings("ignore")

# --- Preparar dados modelo LSTM ---

def normalizar_padronizar_dados(ticker, path_dados_transformados, path_scalers, path_dados_normalizados):
    df = pd.read_csv(path_dados_transformados)

    df = df.sort_values(by=['data', 'hora']).reset_index(drop=True)  # Ordena corretamente por data e hora

    # Seleciona as colunas utilizadas no modelo
    df = df[['data', 'hora', 'dia_semana', 'abertura_candle', 'min_candle', 'max_candle',
             'fechamento_candle', 'volume_candle', 'RSI', 'EMA_20', 'retorno', 'hora_num', 'minuto']]

    # Aplica one-hot encoding na coluna de dia da semana para representar dias como variáveis binárias
    df = pd.get_dummies(df, columns=['dia_semana'], prefix='dia_semana')

    # Define colunas a serem normalizadas (MinMaxScaler) e padronizadas (StandardScaler)
    colunas_normalizar = ['abertura_candle', 'min_candle', 'max_candle', 'fechamento_candle', 'volume_candle']
    colunas_padronizar = ['RSI', 'EMA_20', 'retorno', 'hora_num', 'minuto']

    # Verifica se todas as colunas são numéricas antes da transformação
    for col in colunas_normalizar + colunas_padronizar:
        if not pd.api.types.is_numeric_dtype(df[col]):
            raise ValueError(f"Coluna {col} não é numérica.")

    # Aplica normalização nas colunas definidas
    minmax_scaler = MinMaxScaler()
    df[colunas_normalizar] = minmax_scaler.fit_transform(df[colunas_normalizar])

    # Aplica padronização nas colunas restantes
    standard_scaler = StandardScaler()
    df[colunas_padronizar] = standard_scaler.fit_transform(df[colunas_padronizar])

    # Salva os scalers treinados e metadados com informações úteis para reuso futuro
    os.makedirs(path_scalers, exist_ok=True)
    joblib.dump(minmax_scaler, os.path.join(path_scalers, 'minmax_scaler.pkl'))
    joblib.dump(standard_scaler, os.path.join(path_scalers, 'standard_scaler.pkl'))

    scalers_info = {
        'colunas_normalizadas': colunas_normalizar,
        'colunas_padronizadas': colunas_padronizar,
        'colunas_one_hot': [col for col in df.columns if col.startswith('dia_semana_')]
    }
    joblib.dump(scalers_info, os.path.join(path_scalers, 'scalers_info.pkl'))

    # Salva os dados normalizados em CSV
    os.makedirs(path_dados_normalizados, exist_ok=True)
    df.to_csv(os.path.join(path_dados_normalizados, 'dados_normalizados.csv'), index=False)
    return df

# Função que cria as sequências de entrada (X) e alvo (y) para o modelo LSTM
# Considera dois dias como entrada e um dia como saída (completo de 82 candles)
def criar_sequencias(df, obs_por_dia, target_col, colunas_extra, path_sequencias):
    X, y, extras = [], [], []  # Listas para armazenar entradas, alvos e dados auxiliares
    seq_len = 2 * obs_por_dia  # Número total de observações de entrada (2 dias)
    n_alvos = obs_por_dia      # Número de observações alvo (1 dia)

    # Loop para gerar todas as janelas válidas
    for i in range(len(df) - seq_len - n_alvos + 1):
        seq = df.iloc[i : i + seq_len]  # Entrada
        alvo = df.iloc[i + seq_len : i + seq_len + n_alvos]  # Saída

        datas_alvo = alvo['data'].unique()  # Verifica se o alvo pertence a um único pregão
        if len(datas_alvo) != 1:
            continue

        seq_x = seq.drop(columns=['data', 'hora'] + colunas_extra).values.astype(np.float32)  # Remove colunas extras
        seq_y = alvo[target_col].values.astype(np.float32)  # Extrai valores do alvo
        X.append(seq_x)
        y.append(seq_y)
        extras.append(alvo.iloc[0][colunas_extra].values)  # Armazena info extra do primeiro candle do alvo

    # Converte listas para arrays e salva em arquivos .npy
    X = np.array(X, dtype=np.float32)
    y = np.array(y, dtype=np.float32)
    extras = np.array(extras, dtype=object)

    os.makedirs(path_sequencias, exist_ok=True)
    np.save(os.path.join(path_sequencias, 'X.npy'), X)
    np.save(os.path.join(path_sequencias, 'y.npy'), y)
    np.save(os.path.join(path_sequencias, 'extras.npy'), extras)

    return X, y, extras

# Função que divide os dados em conjuntos de treino, validação e teste conforme proporções definidas
# E salva cada subconjunto em arquivos distintos para uso posterior

def dividir_dados(X, y, extras, path_treino_teste_val, proporcao_treino=0.7, proporcao_val=0.15):
    n = len(X)
    n_treino = int(n * proporcao_treino)
    n_val = int(n * proporcao_val)

    X_treino, y_treino, extras_treino = X[:n_treino], y[:n_treino], extras[:n_treino]
    X_val, y_val, extras_val = X[n_treino:n_treino + n_val], y[n_treino:n_treino + n_val], extras[n_treino:n_treino + n_val]
    X_teste, y_teste, extras_teste = X[n_treino + n_val:], y[n_treino + n_val:], extras[n_treino + n_val:]

    os.makedirs(path_treino_teste_val, exist_ok=True)
    np.save(os.path.join(path_treino_teste_val, 'X_treino.npy'), X_treino)
    np.save(os.path.join(path_treino_teste_val, 'y_treino.npy'), y_treino)
    np.save(os.path.join(path_treino_teste_val, 'extras_treino.npy'), extras_treino)

    np.save(os.path.join(path_treino_teste_val, 'X_val.npy'), X_val)
    np.save(os.path.join(path_treino_teste_val, 'y_val.npy'), y_val)
    np.save(os.path.join(path_treino_teste_val, 'extras_val.npy'), extras_val)

    np.save(os.path.join(path_treino_teste_val, 'X_teste.npy'), X_teste)
    np.save(os.path.join(path_treino_teste_val, 'y_teste.npy'), y_teste)
    np.save(os.path.join(path_treino_teste_val, 'extras_teste.npy'), extras_teste)

    return (X_treino, y_treino, extras_treino,
            X_val, y_val, extras_val,
            X_teste, y_teste, extras_teste)

# Função principal que orquestra o processo de preparação dos dados
def preparar_dados_lstm(ticker, versao_dados, path_projeto):
    print(f"\n--- Preparando dados para o modelo LSTM para {ticker} na versão {versao_dados} ---")

    # Define os caminhos
    path_dados_transformados = os.path.join(path_projeto, f'data/transformed/{ticker}/dados_transformados_{ticker}.csv')
    base_path = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}')
    path_dados_normalizados = os.path.join(base_path, 'dados_normalizados')
    path_scalers = os.path.join(base_path, 'scalers')
    path_sequencias = os.path.join(base_path, 'sequencias')
    path_treino_teste_val = os.path.join(base_path, 'treino_teste_val')

    # Executa a normalização
    print('\n[✔] Normalizando e padronizando dados...')
    df = normalizar_padronizar_dados(ticker, path_dados_transformados, path_scalers, path_dados_normalizados)

    print('\n[✔] Criando sequências de entrada e alvo...')
    obs_por_dia = 82  # Número de candles por dia
    target_col = 'fechamento_candle'  # Coluna alvo para previsão
    colunas_extra = ['data', 'hora']  # Informações auxiliares que serão preservadas separadamente
    # Criação de sequências
    X, y, extras = criar_sequencias(df, obs_por_dia, target_col, colunas_extra, path_sequencias)
    # Divisão dos dados
    print('\n[✔] Dividindo dados em treino, validação e teste...')
    return dividir_dados(X, y, extras, path_treino_teste_val)



#----- Avaliar LSTM e abastecer relatório de acurácia v009 ----------


# Carrega os parametros finais do modelo (ajustados no treinamento)
def carregar_parametros_treinados(path_parametros_treinados):
    with open(path_parametros_treinados, 'rb') as f:
        return pickle.load(f)

# Carrega as métricas resultantes
def carregar_metricas(path_metricas):
    with open(path_metricas, 'r') as f:
        return json.load(f)

# Carrega os dados de teste
def carregar_dados_teste(path_treino_teste_val):
    X_teste = np.load(os.path.join(path_treino_teste_val, 'X_teste.npy'), allow_pickle=True)
    y_teste = np.load(os.path.join(path_treino_teste_val, 'y_teste.npy'), allow_pickle=True)
    return X_teste, y_teste

# Carrega a versao do modelo a ser avaliada
def carregar_modelo(path_modelo):
    return load_model(path_modelo)

# Carrega os scalers usados na transformação
def carregar_scaler_precos(path_scalers):
    return joblib.load(os.path.join(path_scalers, 'minmax_scaler.pkl'))

# Realiza o cálculo das métricas
def calcular_metricas(y_real, y_previsto, nome, path_metricas):
    mae = mean_absolute_error(y_real, y_previsto)
    mse = mean_squared_error(y_real, y_previsto)
    r2 = r2_score(y_real, y_previsto)
    print(f"{nome} - MAE: {mae:.4f}, MSE: {mse:.4f}, R²: {r2:.4f}")

    metricas = {'mae': mae, 'mse': mse, 'r2': r2}
    with open(path_metricas, 'w') as f:
        json.dump(metricas, f, indent=4)
    return mae, mse, r2

# Atualiza ou adiciona o registro de acurácia no relatório csv
# Captura parametros finais ajustados no treino, e evita linhas duplicadas para mesma versão de modelo
def relatorio_acuracia(path_modelo, path_parametros_treinados, path_metricas, path_relatorio_acuracia, versao_modelo, versao_dados):
    parametros_treinados = carregar_parametros_treinados(path_parametros_treinados)
    metricas = carregar_metricas(path_metricas)

    if os.path.exists(path_relatorio_acuracia):
        df_acuracia = pd.read_csv(path_relatorio_acuracia)
        df_acuracia = df_acuracia[df_acuracia['modelo_versao'] != versao_modelo]  # Remove entrada antiga se existir
    else:
        df_acuracia = pd.DataFrame(columns=[
            'modelo_versao', 'dados_versao', 'mae', 'mse', 'r2',
            'look_back_days', 'expected_obs_per_day', 'num_features_por_observacao_intradia',
            'lstm_units_camada1', 'lstm_units_camada2', 'dropout_rate',
            'optimizer', 'learning_rate', 'loss_function',
            'early_stopping_patience', 'max_epochs', 'batch_size',
            'path_projeto'
        ])

    nova_linha = {
        'modelo_versao': versao_modelo,
        'dados_versao': versao_dados,
        'mae': metricas.get('mae'),
        'mse': metricas.get('mse'),
        'r2': metricas.get('r2'),
        'look_back_days': parametros_treinados.get('dias_retrocesso_final'),
        'expected_obs_per_day': parametros_treinados.get('observacoes_por_dia_final'),
        'num_features_por_observacao_intradia': parametros_treinados.get('num_features_por_observacao_final'),
        'lstm_units_camada1': parametros_treinados.get('unidades_lstm_camada1_final'),
        'lstm_units_camada2': parametros_treinados.get('unidades_lstm_camada2_final'),
        'dropout_rate': parametros_treinados.get('taxa_dropout_final'),
        'optimizer': parametros_treinados.get('otimizador_final'),
        'learning_rate': parametros_treinados.get('taxa_aprendizado_final'),
        'loss_function': parametros_treinados.get('funcao_perda_final'),
        'early_stopping_patience': parametros_treinados.get('paciencia_parada_antecipada_final'),
        'max_epochs': parametros_treinados.get('epocas_maximas_final'),
        'batch_size': parametros_treinados.get('tamanho_lote_final'),
        'path_projeto': parametros_treinados.get('path_projeto_final')
    }

    df_acuracia.loc[len(df_acuracia)] = nova_linha
    df_acuracia.to_csv(path_relatorio_acuracia, index=False)

# Realiza a plotagem da comparação entre o pregão real e o pregão previsto
def plotar_comparacao_previsoes(y_real, y_previsto, obs_por_dia, titulo_prefixo="", max_dias_plot=6):
    if y_real.shape != y_previsto.shape:
        raise ValueError("Os shapes de y_real e y_previsto devem ser idênticos.")

    num_dias = min(y_real.shape[0] // obs_por_dia, max_dias_plot)
    if num_dias == 0:
        print("Nenhum dado para plotar.")
        return

    print(f"\n--- Gerando {num_dias} gráficos de comparação (real vs. previsto) ---")
    fig, axes = plt.subplots(nrows=num_dias, ncols=1, figsize=(18, 5 * num_dias))
    if num_dias == 1:
        axes = [axes]

    for i in range(num_dias):
        ax = axes[i]
        start = i * obs_por_dia
        end = start + obs_por_dia
        indices = np.arange(obs_por_dia)

        ax.plot(indices, y_real[start:end], label='Real', color='#1f77b4', linewidth=2)
        ax.plot(indices, y_previsto[start:end], label='Previsto', color='#d62728', linestyle='--', linewidth=2)

        ax.set_title(f'Dia {i+1} - {titulo_prefixo} - Previsão de Fechamento (Real vs. Previsto)', fontsize=14)
        ax.set_xlabel('Índice do Candle (5 minutos)', fontsize=12)
        ax.set_ylabel('Fechamento (Desnormalizado)', fontsize=12)
        ax.legend(fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.7)

    plt.tight_layout(rect=[0, 0.03, 1, 0.98])
    plt.show()

# Função principal de avaliação
def avaliar_lstm(ticker, versao_modelo, versao_dados, path_projeto):
    path_modelo_pasta = os.path.join(path_projeto, f'models/LSTM/{ticker}/modelo_lstm_{versao_modelo}')
    path_modelo = os.path.join(path_modelo_pasta, f'modelo_lstm_{versao_modelo}.h5')
    path_parametros_treinados = os.path.join(path_modelo_pasta, f'parametros_{versao_modelo}.pkl')
    path_metricas = os.path.join(path_modelo_pasta, f'metricas_{versao_modelo}.json')
    path_treino_teste_val = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}/treino_teste_val')
    path_scalers = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}/scalers')
    path_relatorio_acuracia = os.path.join(path_projeto, f'models/LSTM/{ticker}/relatorio_acuracia.csv')

    print("Carregando dados de teste...")
    X_teste, y_teste = carregar_dados_teste(path_treino_teste_val)

    print("Carregando modelo...")
    modelo = carregar_modelo(path_modelo)

    print("Carregando scaler de preços...")
    scaler_precos = carregar_scaler_precos(path_scalers)

    print("Realizando previsões...")
    y_previsto = modelo.predict(X_teste)

    colunas_precos = ['abertura_candle', 'min_candle', 'max_candle', 'fechamento_candle', 'volume_candle']
    idx_fechamento = colunas_precos.index('fechamento_candle')

    print("Desnormalizando previsões...")
    y_previsto_inv = scaler_precos.inverse_transform(
        np.pad(y_previsto.reshape(-1, 1),
               ((0, 0), (idx_fechamento, len(colunas_precos) - idx_fechamento - 1)),
               constant_values=0)
    )[:, idx_fechamento].reshape(y_previsto.shape)

    y_teste_inv = scaler_precos.inverse_transform(
        np.pad(y_teste.reshape(-1, 1),
               ((0, 0), (idx_fechamento, len(colunas_precos) - idx_fechamento - 1)),
               constant_values=0)
    )[:, idx_fechamento].reshape(y_teste.shape)

    print("Calculando métricas...")
    calcular_metricas(y_teste_inv.flatten(), y_previsto_inv.flatten(), "Fechamento", path_metricas)

    print("Gerando relatório de acurácia...")
    relatorio_acuracia(path_modelo, path_parametros_treinados, path_metricas, path_relatorio_acuracia, versao_modelo, versao_dados)

    print("Gerando gráfico de comparação...")
    obs_por_dia = y_teste.shape[1]
    plotar_comparacao_previsoes(y_teste_inv.flatten(), y_previsto_inv.flatten(), obs_por_dia, titulo_prefixo="LSTM", max_dias_plot=6)

    comparacao = pd.DataFrame({
        'Fechamento_Real': y_teste_inv.flatten(),
        'Fechamento_Previsto': y_previsto_inv.flatten()
    })

    return y_teste_inv, y_previsto_inv, comparacao, carregar_parametros_treinados(path_parametros_treinados)


#  --- Treinar modelo LSTM ---

def treinar_lstm(ticker, versao_modelo, versao_dados, path_projeto, parametros: dict):
    print("Iniciando treinamento do modelo LSTM...")

    # Define os caminhos
    path_modelo_pasta = os.path.join(path_projeto, f'models/LSTM/{ticker}/modelo_lstm_{versao_modelo}')
    path_modelo = os.path.join(path_modelo_pasta, f'modelo_lstm_{versao_modelo}.h5')
    path_parametros_modelo = os.path.join(path_modelo_pasta, f'parametros_{versao_modelo}.pkl')
    path_treino_teste_val = os.path.join(path_projeto, f'data/prepared/{ticker}/{versao_dados}/treino_teste_val')

    os.makedirs(path_modelo_pasta, exist_ok=True)

    # Carregar dados
    X_treino = np.load(os.path.join(path_treino_teste_val, 'X_treino.npy'), allow_pickle=True)
    y_treino = np.load(os.path.join(path_treino_teste_val, 'y_treino.npy'), allow_pickle=True)
    X_val = np.load(os.path.join(path_treino_teste_val, 'X_val.npy'), allow_pickle=True)
    y_val = np.load(os.path.join(path_treino_teste_val, 'y_val.npy'), allow_pickle=True)

    # Ajustar dimensões
    if len(X_treino.shape) == 2:
        X_treino = X_treino.reshape((X_treino.shape[0], X_treino.shape[1], 1))
        X_val = X_val.reshape((X_val.shape[0], X_val.shape[1], 1))

    # Verificação para garantir que tenhamos multiplas saidas
    if len(y_treino.shape) == 1:
        y_treino = y_treino.reshape((y_treino.shape[0], -1))
        y_val = y_val.reshape((y_val.shape[0], -1))

    # Configurar otimizador
    otimizador_nome = parametros['otimizador'].lower()
    taxa_aprendizado = parametros['taxa_aprendizado']
    if otimizador_nome == 'adam':
        otimizador = Adam(learning_rate=taxa_aprendizado)
    else:
        raise ValueError(f"Otimizador '{otimizador_nome}' não implementado.")

    # Configurar função perda
    perda_nome = parametros['funcao_perda'].lower()
    if perda_nome == 'huber':
        funcao_perda = Huber()
    else:
        raise ValueError(f"Função de perda '{perda_nome}' não implementada.")

    # Construção do modelo
    modelo = Sequential([
        LSTM(parametros['unidades_lstm_camada1'], return_sequences=True,
             input_shape=(X_treino.shape[1], X_treino.shape[2])),
        Dropout(parametros['taxa_dropout']),
        LSTM(parametros['unidades_lstm_camada2'], return_sequences=False),
        Dropout(parametros['taxa_dropout']),
        Dense(y_treino.shape[1])  # Saída com tamanho igual ao número de candles do pregão
    ])

    modelo.compile(optimizer=otimizador, loss=funcao_perda)

    # Callbacks
    callbacks = [
        EarlyStopping(monitor='val_loss', patience=parametros['paciencia_parada_antecipada'], restore_best_weights=True),
        ModelCheckpoint(path_modelo, monitor='val_loss', save_best_only=True)
    ]

    # Treinamento
    historico = modelo.fit(
        X_treino, y_treino,
        validation_data=(X_val, y_val),
        epochs=parametros['epocas_maximas'],
        batch_size=parametros['tamanho_lote'],
        callbacks=callbacks,
        verbose=1
    )

    print(f"[✓] Modelo treinado e salvo em: {path_modelo}")
    print(f"Total de épocas executadas: {len(historico.history['loss'])}")

    # Captura real dos parâmetros finais a partir do modelo
    config_modelo = modelo.get_config()
    parametros_modelo = {
        'ticker': ticker,
        'versao': versao_modelo,
        'path_modelo': path_modelo,
        'path_projeto': path_projeto,
        'epocas_executadas': len(historico.history['loss']),
        'loss_final_treino': historico.history['loss'][-1],
        'loss_final_val': historico.history['val_loss'][-1],
        'layers': [layer['class_name'] for layer in config_modelo['layers']]
    }

    print("[✓] Parâmetros finais do modelo:")
    for chave, valor in parametros_modelo.items():
        print(f"  {chave}: {valor}")

    with open(path_parametros_modelo, 'wb') as f:
        pickle.dump(parametros_modelo, f)

    print(f"[✓] Parâmetros de treinamento salvos em: {path_parametros_modelo}")

    return modelo, parametros_modelo

# Execução direta para testes
if __name__ == '__main__':
    parametros = {
        "dias_retrocesso": 2,
        "observacoes_por_dia": 82,
        "num_features_por_observacao": 20,
        "unidades_lstm_camada1": 128,
        "unidades_lstm_camada2": 128,
        "taxa_dropout": 0.25,
        "otimizador": "Adam",
        "taxa_aprendizado": 0.006,
        "funcao_perda": "Huber",
        "paciencia_parada_antecipada": 25,
        "epocas_maximas": 300,
        "tamanho_lote": 32
    }
    ticker = 'BBDC4'
    versao_modelo = 'vteste' # Dados preparados anterioremnte
    versao_dados = 'dados_teste'
    path_projeto = '/content/Piloto_day_trade'

    # preparar_dados_lstm(ticker, versao_dados, path_projeto)
    preparar_dados_lstm(ticker, versao_dados, path_projeto)
    treinar_lstm(ticker, versao_modelo, versao_dados, path_projeto, parametros)
    avaliar_lstm(ticker, versao_modelo, versao_dados, path_projeto)


```
